```python
#importing libraries
import pandas as pd
import numpy as np
```

```python
#reading cafe sales csv
df= pd.read_csv('/workspaces/Data_Cleaning_Project/data/raw/Cafe Sales.csv')
```

```python
#inspecting first 10 records
df.head(10)
```

```python
#checking the basic structure and data types
df.info()
```

```python
#checking if there are any duplicate records
print(f'This dataset contains {sum(df.duplicated())} duplicate records.')
```

```python
#inspecting missing data for each column
df.isna().sum()* 100/df.shape[0]
```

```python
#inspecting unique values in all columns except Transaction ID
df_cols=set(df.columns)
exclude={'Transaction ID'}
df_cols=df_cols.difference(exclude)
df_cols=list(df_cols)
for col in df_cols:
    print(f'{col}:\n {df[col].unique()}' , end = '\n\n')
```

```python
df_cols
```

```python
#replacing ERROR and UNKNOWN with nan
df=df.replace(['ERROR','UNKNOWN'], np.nan)
```

```python
#changing data types
df['Quantity']=df['Quantity'].astype('Int64')
df['Price Per Unit']= df['Price Per Unit'].astype('float')
df['Total Spent']=df['Total Spent'].astype('float')
df['Transaction Date']=pd.to_datetime(df['Transaction Date'])
```

```python
#rechecking data types
df.info()
```

```python
#rechecking missing data
df.isna().sum() *100 /df.shape[0]
```

```python
#droping records where transaction date is not available
df=df.dropna(subset='Transaction Date')
```

```python
#replacing null with 'INA'(information not available) to avoid bais in Payment method and location
df['Payment Method']= df['Payment Method'].fillna('INA')
df['Location']= df['Location'].fillna('INA')
```

```python
#replacing null quantities with mean values
quantity_mean= round(df['Quantity'].mean(),0)
df['Quantity']= df['Quantity'].fillna(quantity_mean)
```

```python
#replacing null values in item with most frequent values
item_mode=df['Item'].mode()[0]
df['Item']=df['Item'].fillna(item_mode)
```

```python
#checking price per unit values with most frequent price for each category
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
#filling missing values in price per unit with most fequent price for each category
df['Price Per Unit']=df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
# calculating Total spent 
df['Total Spent']=df['Quantity']*df['Price Per Unit']
```

```python
#rechecking missing data
df.isna().sum()*100/df.shape[0]
```

```python
df.to_csv('/workspaces/Data_Cleaning_Project/data/clean/Cafe Sales (Clean).csv', index= False)
```
